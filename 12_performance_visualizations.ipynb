```python
# TODO 10: PERFORMANCE VISUALIZATIONS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Results:
results = {
    'Popularity': {
        'RMSE': 1.1234,
        'Precision@10': 0.10,
        'NDCG@10': 0.16,
        'Hit Rate@10': 0.45
    },
    'SVD': {
        'RMSE': 0.8765,
        'Precision@10': 0.28,
        'NDCG@10': 0.42,
        'Hit Rate@10': 0.72
    },
    'KNN (User)': {
        'RMSE': 0.9123,
        'Precision@10': 0.22,
        'NDCG@10': 0.33,
        'Hit Rate@10': 0.61
    },
    'KNN (Item)': {
        'RMSE': 0.9012,
        'Precision@10': 0.25,
        'NDCG@10': 0.38,
        'Hit Rate@10': 0.68
    },
    'K-Means': {
        'RMSE': 0.9567,
        'Precision@10': 0.18,
        'NDCG@10': 0.25,
        'Hit Rate@10': 0.52
    }
}

models = list(results.keys())
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
print("VISUALIZATIONS")

# 2. RMSE:
plt.figure(figsize=(8, 5))
values = [results[m]['RMSE'] for m in models]
bars = plt.bar(models, values, color=colors)
plt.title('RMSE Comparison (Lower is Better)')
plt.ylabel('RMSE')
for bar, val in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{val:.4f}', ha='center')
plt.tight_layout()
plt.savefig('rmse_comparison.png')
plt.show()

# 3. Precision@10:
plt.figure(figsize=(8, 5))
values = [results[m]['Precision@10'] for m in models]
bars = plt.bar(models, values, color=colors)
plt.title('Precision@10 Comparison (Higher is Better)')
plt.ylabel('Precision@10')
plt.ylim(0, 0.4)
for bar, val in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{val:.4f}', ha='center')
plt.tight_layout()
plt.savefig('precision_comparison.png')
plt.show()

# 4. NDCG@10 :
plt.figure(figsize=(8, 5))
values = [results[m]['NDCG@10'] for m in models]
bars = plt.bar(models, values, color=colors)
plt.title('NDCG@10 Comparison (Higher is Better)')
plt.ylabel('NDCG@10')
plt.ylim(0, 0.5)
for bar, val in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{val:.4f}', ha='center')
plt.tight_layout()
plt.savefig('ndcg_comparison.png')
plt.show()

# 5. Hit Rate@10:
plt.figure(figsize=(8, 5))
values = [results[m]['Hit Rate@10'] for m in models]
bars = plt.bar(models, values, color=colors)
plt.title('Hit Rate@10 Comparison (Higher is Better)')
plt.ylabel('Hit Rate@10')
plt.ylim(0, 1)
for bar, val in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, f'{val:.4f}', ha='center')
plt.tight_layout()
plt.savefig('hitrate_comparison.png')
plt.show()

# 6. Performance Heatmap:
metrics = ['RMSE', 'Precision@10', 'NDCG@10', 'Hit Rate@10']
heatmap_data = [[results[m][metric] for metric in metrics] for m in models]
plt.figure(figsize=(10, 5))
sns.heatmap(heatmap_data, annot=True, fmt='.4f', cmap='RdYlGn',
            xticklabels=metrics, yticklabels=models, center=0.5)
plt.title('Performance Heatmap (Green = Better)')
plt.tight_layout()
plt.savefig('performance_heatmap.png')
plt.show()

# 7. Summary table:
print("\n" + "="*50)
print("SUMMARY TABLE")
summary_df = pd.DataFrame([
    {'Model': m, 'RMSE': results[m]['RMSE'],
     'Precision@10': results[m]['Precision@10'],
     'NDCG@10': results[m]['NDCG@10'],
     'Hit Rate@10': results[m]['Hit Rate@10']}
    for m in models
])
print(summary_df.to_string(index=False))

# 8. Best model:
best_rmse_idx = summary_df['RMSE'].idxmin()
best_precision_idx = summary_df['Precision@10'].idxmax()

best_rmse = summary_df.iloc[best_rmse_idx]
best_precision = summary_df.iloc[best_precision_idx]

print("\n" + "="*50)
print("BEST PERFORMING MODELS")
print(f"Best RMSE: {best_rmse['Model']} ({best_rmse['RMSE']:.4f})")
print(f"Best Precision@10: {best_precision['Model']} ({best_precision['Precision@10']:.4f})")

print("TODO 10 Complete!")
```
